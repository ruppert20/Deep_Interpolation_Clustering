```python
import numpy as np
import pandas as pd
```

```python
# Load final cluster assignments
data = np.load('Results/Clustering/out_feat/ae_mse_kmeans_aligned/training_3.npy',
               allow_pickle=True).item()

encounter_ids = data['encounter_id'].astype(str)
features = data['hidden']
cluster_ids = data['cluster_id']

# Count samples per cluster
unique, counts = np.unique(cluster_ids, return_counts=True)
for c, n in zip(unique, counts):
    print(f"Cluster {c}: {n} samples")
```

```python
# Create DataFrame for analysis
df = pd.DataFrame({
    'encounter_id': data['encounter_id'].astype(str),
    'cluster': data['cluster_id']
})

print(df.dtypes)

# Merge with outcomes data for analysis
outcomes = pd.read_csv('Data/analysis_data/table_data.csv').rename(columns={'encounter_deiden_id': 'encounter_id'}, inplace=False)
outcomes['encounter_id'] = outcomes['encounter_id'].astype(str)
print(outcomes.head(4))
analysis = df.merge(outcomes, on='encounter_id')

# print(analysis.head(4))

# Compare outcomes by cluster
print('Combined Endpoint Distribution by Cluster:')
print(analysis.groupby('cluster')['combined_endpoint'].value_counts(normalize=True))

print('ICU Distribution by Cluster:')
print(analysis.groupby('cluster')['ICU'].value_counts(normalize=True))

print('Rapid Response Distribution by Cluster:')
print(analysis.groupby('cluster')['rapid_response'].value_counts(normalize=True))
```
